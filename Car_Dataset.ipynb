```python
import pandas as pd

# Read the Excel file
df = pd.read_excel(r"E:\Downloads\Dataset\Car_Sales.xlsx")

# Print the columns of the DataFrame to find the correct column name
print(df.columns)

# Assuming the date column name is 'Date' (replace 'Date' with the actual column name if different)
df['Date'] = pd.to_datetime(df['Date'])

# Convert to the desired format
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')

# Print the dataframe to verify changes
print(df)
```
